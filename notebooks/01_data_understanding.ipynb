```python
numeric_columns = [
    "price",
    "positive",
    "negative",
    "average_playtime_forever"
]

print(df[numeric_columns].describe())
```

```python
# Negative prices
print("Negative Prices:")
print(df[df["price"].astype("string").str.contains("-", na=False)])

# Negative review counts
print("Negative Reviews:")
print(df[
    (pd.to_numeric(df["positive"], errors="coerce") < 0) |
    (pd.to_numeric(df["negative"], errors="coerce") < 0)
])

# Negative playtime
print("Negative Playtime:")
print(df[
    pd.to_numeric(df["average_playtime_forever"], errors="coerce") < 0
])
```

```python
for col in numeric_columns:

    values = pd.to_numeric(df[col], errors="coerce").dropna()

    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = values[
        (values < lower_limit) |
        (values > upper_limit)
    ]

    print("\nColumn:", col)
    print("Potential Outliers:", len(outliers))
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
```

```python
quality_report = pd.DataFrame({
    "Check": [
        "Total Rows",
        "Total Columns",
        "Missing Values",
        "Duplicate Records",
        "Unique Game Names",
        "Missing Game Names",
        "Negative Prices",
        "Negative Review Counts",
        "Negative Playtime"
    ],

    "Result": [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df["name"].nunique(),
        df["name"].isnull().sum(),
        (
            pd.to_numeric(df["price"], errors="coerce") < 0
        ).sum(),
        (
            (pd.to_numeric(df["positive"], errors="coerce") < 0) |
            (pd.to_numeric(df["negative"], errors="coerce") < 0)
        ).sum(),
        (
            pd.to_numeric(
                df["average_playtime_forever"],
                errors="coerce"
            ) < 0
        ).sum()
    ]
})

print(quality_report)
```

```python
quality_report.to_csv(
    "../reports/steam_data_quality_report.csv",
    index=False
)
```
